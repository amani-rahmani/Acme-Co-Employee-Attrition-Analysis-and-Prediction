```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

#info about the dataset
dataset = pd.read_csv('/content/Employee-Attrition.csv')
# number of rows and columns in the dataframe
dataset.shape

```

```python
#info about the dataset
dataset.info()

```

```python
#stastical info about the data
dataset.describe()
```

```python
#checking if theres any missing values
dataset.isnull().sum()
```

```python
# Droping rows with missing values
dataset.dropna(inplace=True)
```

```python
# Change Factors to Numerics
dataset['Attrition'] = dataset['Attrition'].apply (lambda x: 1 if x == "Yes" else 0)
dataset['OverTime'] = dataset['OverTime'].apply (lambda x: 1 if x == "Yes" else 0)
```

```python

attrition_count = pd.DataFrame(dataset['Attrition'].value_counts())
attrition_count
```

```python
attrition_rate = dataset["Attrition"].mean()
print(f"Attrition Rate: {attrition_rate:.2%}")
```

```python
sns.countplot(x="Attrition", data=dataset,color='red')
plt.xlabel("Attrition")
plt.ylabel("Number of Employees" )
plt.title("Overall Employee Attrition")
plt.show()
```

```python
sns.countplot(x="OverTime", hue="Attrition", data=dataset,palette='dark:red')
plt.title("Attrition by Overtime")
plt.show()
```

```python
sns.boxplot(x="Attrition", y="MonthlyIncome", data=dataset,color='red')
plt.title("Monthly Income vs Attrition")
plt.show()
```

```python
sns.boxplot(x="Attrition", y="JobSatisfaction", data=dataset,color='red')
plt.title("Job Satisfaction vs Attrition")
plt.show()
```

```python
sns.boxplot(
    x="Attrition",
    y="WorkLifeBalance",
    data=dataset,
    color = 'red'
)
plt.title("Work-Life Balance vs Attrition")
plt.show()
```

```python
sns.boxplot(
    x="Attrition",
    y="Age",
    data=dataset,
    color='red'
)
plt.title("Age vs Attrition")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15,5))

sns.boxplot(x="Attrition", y="YearsAtCompany", data=dataset, ax=axes[0],color='red')
sns.boxplot(x="Attrition", y="YearsInCurrentRole", data=dataset, ax=axes[1],color='red')
sns.boxplot(x="Attrition", y="YearsSinceLastPromotion", data=dataset, ax=axes[2],color='red')

axes[0].set_title("Years at Company")
axes[1].set_title("Years in Current Role")
axes[2].set_title("Years Since Last Promotion")

plt.tight_layout()
plt.show()
```

```python
satisfaction_cols = [
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction"
]

for col in satisfaction_cols:
    sns.boxplot(x="Attrition", y=col, data=dataset,color='red')
    plt.title(f"{col} vs Attrition")
    plt.show()
```
